```python
# Task 1: Mean of daily step counts

daily_steps = [8214, 6532, 10023, 7456, 9182, 5890, 11340]

def calculate_mean(data):
    return sum(data) / len(data)

mean_steps = calculate_mean(daily_steps)
print("Task 1 - Daily steps:", daily_steps)
print("Task 1 - Mean steps:", round(mean_steps, 2))
```

```python
# Task 2: Median of Zomato/Swiggy delivery times

delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2
    return sorted_data[mid]

median_time = find_median(delivery_times)
print("\nTask 2 - Delivery times:", delivery_times)
print("Task 2 - Median delivery time:", median_time, "minutes")
```

```python
# Task 3: Mode of YouTube genres watched

genres = ['Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog', 'Tech',
          'Music', 'Comedy', 'Vlog', 'Music', 'Comedy']

def find_mode(data):
    counts = {}
    for item in data:
        counts[item] = counts.get(item, 0) + 1
    max_count = max(counts.values())
    modes = [k for k, v in counts.items() if v == max_count]
    return modes, counts

modes, genre_counts = find_mode(genres)
print("\nTask 3 - Genre counts:", genre_counts)
print("Task 3 - Mode (most watched genre):", modes[0], f"({genre_counts[modes[0]]} times)")
```

```python
# Task 4: Mean vs median of UPI transaction amounts

upi_amounts = [100, 120, 105, 110, 5000, 115, 108]

mean_amount = calculate_mean(upi_amounts)
median_amount = find_median(upi_amounts)

print("\nTask 4 - UPI amounts:", upi_amounts)
print("Task 4 - Mean:", round(mean_amount, 2))
print("Task 4 - Median:", median_amount)

explanation = """
The median (Rs. 110) is the better measure of a 'typical' transaction here.
The single Rs. 5000 outlier pulls the mean up to Rs. 808.29, far above what
most transactions actually look like, while the median stays close to the
bulk of the data since it only depends on the middle value's position."""
```
