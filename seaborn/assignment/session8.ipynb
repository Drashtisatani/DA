```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style='whitegrid')
np.random.seed(6)
```

## Task 1 — Scatterplot: Instagram Followers vs Posts (20 Users)

```python
n = 20
posts = np.random.randint(20, 500, n)
followers = (posts * np.random.uniform(15, 80, n) + np.random.normal(0, 2000, n)).clip(200, None)
insta_users = pd.DataFrame({'posts': posts, 'followers': followers.astype(int)})

plt.figure(figsize=(7, 4.5))
sns.scatterplot(data=insta_users, x='posts', y='followers')
plt.xlabel('Number of Posts')
plt.ylabel('Number of Followers')
plt.title('Instagram: Followers vs Posts (20 Fictional Users)')
plt.tight_layout()
plt.show()
```

## Task 2 — Lineplot: Daily Zomato Orders Over a Month

```python
days = np.arange(1, 31)
daily_orders = np.random.randint(150, 400, 30)
orders_month_df = pd.DataFrame({'day': days, 'orders': daily_orders})

plt.figure(figsize=(9, 4.5))
sns.lineplot(data=orders_month_df, x='day', y='orders', marker='o')
plt.xlabel('Day of Month')
plt.ylabel('Number of Orders')
plt.title('Daily Zomato Orders Over a Month')
plt.tight_layout()
plt.show()
```

## Task 3 — Relplot: Movie Ratings vs Reviews by Genre (15 Movies)

```python
genres = ['Action'] * 5 + ['Comedy'] * 5 + ['Drama'] * 5
ratings = np.concatenate([
    np.random.uniform(6.5, 8.5, 5),
    np.random.uniform(6.0, 8.0, 5),
    np.random.uniform(7.0, 9.0, 5),
]).round(1)
reviews = np.random.randint(500, 20000, 15)

movies_df = pd.DataFrame({
    'movie': [f'Movie_{i}' for i in range(1, 16)],
    'rating': ratings,
    'reviews': reviews,
    'genre': genres,
})

g = sns.relplot(data=movies_df, x='rating', y='reviews', hue='genre', s=100, height=5, aspect=1.3)
g.set_axis_labels('Rating', 'Number of Reviews')
g.figure.suptitle('Movie Ratings vs Reviews by Genre', y=1.03)
plt.show()
```

## Task 4 — Regplot: YouTube Hours vs Exam Scores (25 Students)

```python
n4 = 25
youtube_hours = np.random.uniform(0, 6, n4)
exam_scores = (85 - youtube_hours * 6 + np.random.normal(0, 6, n4)).clip(30, 100)
students_df = pd.DataFrame({'youtube_hours': youtube_hours, 'exam_score': exam_scores})

plt.figure(figsize=(7, 4.5))
sns.regplot(data=students_df, x='youtube_hours', y='exam_score')
plt.xlabel('Hours Spent on YouTube (daily)')
plt.ylabel('Exam Score')
plt.title('YouTube Hours vs Exam Scores')
plt.tight_layout()
plt.show()

corr = students_df['youtube_hours'].corr(students_df['exam_score'])
print('Correlation:', round(corr, 3))
```

**Interpretation:** the trend is clearly **negative** — the regression line slopes downward, and the correlation coefficient printed above confirms it. In this (simulated) group, students who spend more daily hours on YouTube tend to score lower on exams, though this shows an association, not proof that YouTube time itself is the cause.
