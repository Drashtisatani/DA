```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme(style='whitegrid')
np.random.seed(2)
```

## Task 1 — Histplot: Swiggy Delivery Times (35 values)

```python
delivery_times = np.random.normal(32, 7, 35).round(1)
delivery_times = np.clip(delivery_times, 12, None)

plt.figure(figsize=(7, 4.5))
sns.histplot(delivery_times, bins=10)
plt.xlabel('Delivery Time (min)')
plt.ylabel('Count')
plt.title('Distribution of Swiggy Delivery Times')
plt.tight_layout()
plt.show()
```

## Task 2 — Kdeplot: One Month of Daily Step Counts

```python
daily_steps = np.random.normal(7500, 1500, 30).round().astype(int)
daily_steps = np.clip(daily_steps, 2000, None)

plt.figure(figsize=(7, 4.5))
sns.kdeplot(daily_steps, fill=True)
plt.xlabel('Daily Steps')
plt.ylabel('Density')
plt.title('Probability Density of Daily Step Counts (1 Month)')
plt.tight_layout()
plt.show()
```

## Task 3 — Histplot + Kdeplot Overlay: Zomato Monthly Spending

```python
monthly_spend = np.random.gamma(shape=3, scale=600, size=100).round()

plt.figure(figsize=(7, 4.5))
sns.histplot(monthly_spend, bins=20, stat='density', color='#4c9aff', alpha=0.5)
sns.kdeplot(monthly_spend, color='red', linewidth=2)
plt.xlabel('Monthly Spending on Zomato (Rs.)')
plt.ylabel('Density')
plt.title('Monthly Zomato Spending: Histogram + KDE Overlay')
plt.tight_layout()
plt.show()
```

## Task 4 — IPL Run Distribution: Varying Bin Count

```python
ipl_scores = np.random.normal(155, 30, 200).round().astype(int)
ipl_scores = np.clip(ipl_scores, 60, 260)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, b in zip(axes, [5, 10, 20]):
    sns.histplot(ipl_scores, bins=b, ax=ax)
    ax.set_title(f'bins={b}')
    ax.set_xlabel('Runs')
    ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

**Observation:** at `bins=5`, the histogram is too coarse — it flattens the distribution into a few wide blocks and hides the actual peak location. At `bins=20`, the shape becomes visibly noisier and jagged, since some bins end up with very few scores by chance, making it harder to judge the underlying trend. `bins=10` strikes the best balance here — enough resolution to see the roughly bell-shaped spread around ~155 runs without the bin-to-bin noise that `bins=20` introduces.
