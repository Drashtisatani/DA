## Task 1 — Sample Space of a Die Roll + Coin Flip

The die has 6 outcomes and the coin has 2, so the combined experiment has 6 × 2 = 12 equally likely outcomes.

```python
die_faces = [1, 2, 3, 4, 5, 6]
coin_sides = ['H', 'T']

sample_space = list(itertools.product(die_faces, coin_sides))

print('Sample space (die, coin):')
print(sample_space)
print('Total outcomes:', len(sample_space))
```

## Task 2 — Classical Probability of Drawing an Even-Numbered Card

Cards numbered 1–10 → 10 total outcomes.
Even numbers: 2, 4, 6, 8, 10 → 5 favorable outcomes.

$$P(\text{even}) = \dfrac{\text{favorable outcomes}}{\text{total outcomes}} = \dfrac{5}{10}$$

```python
cards = list(range(1, 11))
even_cards = [c for c in cards if c % 2 == 0]

p_even = len(even_cards) / len(cards)

print('Cards:', cards)
print('Even cards:', even_cards)
print('P(even) =', len(even_cards), '/', len(cards), '=', p_even)
```

## Task 3 — Probability a Random Song is NOT by Arijit Singh

Playlist has 8 songs, 3 by Arijit Singh, so 5 are by other artists.

$$P(\text{Arijit}) = \dfrac{3}{8} \quad\Rightarrow\quad P(\text{not Arijit}) = 1 - P(\text{Arijit}) = 1 - \dfrac{3}{8} = \dfrac{5}{8}$$

```python
total_songs = 8
arijit_songs = 3
non_arijit_songs = total_songs - arijit_songs

p_arijit = arijit_songs / total_songs
p_not_arijit = 1 - p_arijit

print('P(Arijit) =', arijit_songs, '/', total_songs, '=', p_arijit)
print('P(not Arijit) = 1 -', p_arijit, '=', p_not_arijit)
```

## Task 4 — Addition Rule: P(Pizza or Burger)

$$P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)$$

```python
p_pizza = 0.3
p_burger = 0.4
p_both = 0.1

p_pizza_or_burger = p_pizza + p_burger - p_both

print('P(Pizza or Burger) =', p_pizza, '+', p_burger, '-', p_both, '=', p_pizza_or_burger)
```

## Task 5 — Conditional Probability: Phone Add-to-Cart

```python
p_phone = 0.2
p_phone_given_mobile_interest = 0.5

print('P(add phone to cart) =', p_phone)
print('P(add phone to cart | interested in mobiles) =', p_phone_given_mobile_interest)
```

**Explanation:** Without any extra information, only 20% of Flipkart users add a phone to cart. But once you know a user is specifically browsing mobiles, that probability jumps to 50%. That's what conditional probability captures: narrowing the population down to a relevant subgroup ("interested in mobiles") changes the odds of the event ("adds a phone to cart"), because knowing they're already mobile-focused makes a phone purchase far more likely than it is for a random, unfiltered visitor.
