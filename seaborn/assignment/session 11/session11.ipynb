````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import os

os.makedirs('exports', exist_ok=True)
sns.set_theme(context='notebook', style='whitegrid')
np.random.seed(9)
````

## Task 1

````python
food_categories = pd.DataFrame({
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [420, 350, 510, 290, 180],
})

plt.figure(figsize=(7, 4.5))
sns.barplot(data=food_categories, x='category', y='orders', hue='category', legend=False)
plt.xlabel('Food Category')
plt.ylabel('Number of Orders')
plt.title('Orders per Food Category')
plt.tight_layout()
plt.savefig('exports/orders_by_category_300dpi.png', dpi=300)
plt.show()

print('Saved: exports/orders_by_category_300dpi.png (300 DPI)')
````

## Task 2 — Line Plot Exported as SVG

````python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps_week = [7200, 8100, 6500, 9000, 7800, 10200, 6900]
steps_week_df = pd.DataFrame({'day': days, 'steps': steps_week})

plt.figure(figsize=(7, 4.5))
sns.lineplot(data=steps_week_df, x='day', y='steps', marker='o')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.title('Weekly Step Count')
plt.tight_layout()
plt.savefig('exports/weekly_steps.svg', format='svg')
plt.show()

print('Saved: exports/weekly_steps.svg (vector, PowerPoint-ready)')
````

## Task 3 — Darkgrid Chart Exported as 1200 DPI PNG

````python
sns.set_theme(style='darkgrid', palette='mako')

quarterly_sales = pd.DataFrame({
    'quarter': ['Q1', 'Q2', 'Q3', 'Q4'],
    'sales': [125000, 138000, 119000, 156000],
})

plt.figure(figsize=(7, 4.5))
sns.barplot(data=quarterly_sales, x='quarter', y='sales', hue='quarter', legend=False)
plt.xlabel('Quarter')
plt.ylabel('Sales (Rs.)')
plt.title('Quarterly Sales (darkgrid, mako palette)')
plt.tight_layout()
plt.savefig('exports/quarterly_sales_1200dpi.png', dpi=1200)
plt.show()

sns.set_theme(context='notebook', style='whitegrid')  # reset
print('Saved: exports/quarterly_sales_1200dpi.png (1200 DPI)')
````

## Task 4 — apply_presentation_theme() Template Function + Pie Chart Export

````python
def apply_presentation_theme():
    sns.set_theme(
        context='talk',
        style='white',
        palette='deep',
        font_scale=1.0,
    )
    plt.rcParams['font.family'] = 'sans-serif'

apply_presentation_theme()

apps = ['WhatsApp', 'Instagram', 'YouTube', 'Chrome', 'Spotify']
usage_hours = [18, 14, 12, 9, 7]

plt.figure(figsize=(6.5, 6.5))
colors = sns.color_palette('deep', len(apps))
plt.pie(usage_hours, labels=apps, autopct='%1.1f%%', colors=colors, startangle=90)
plt.title('Top 5 Most-Used Apps (Weekly Hours)')
plt.tight_layout()
plt.savefig('exports/top5_apps_pie.png', dpi=300)
plt.show()

sns.set_theme(context='notebook', style='whitegrid')  # reset
print('Saved: exports/top5_apps_pie.png')
````

**Adapted and run below** on a small sample DataFrame (SVG is vector-based, so `dpi` doesn't change file quality the way it does for PNG — it's included here to match what the snippet specified, but note below on why it's largely a no-op for SVG):

````python
sample_df = pd.DataFrame({
    'delivery_time': np.random.normal(30, 6, 50),
    'order_value': np.random.normal(400, 100, 50),
    'rating': np.random.normal(4.2, 0.4, 50),
})
corr = sample_df.corr()

plt.figure(figsize=(6.5, 5.5))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('exports/heatmap_1200dpi.svg', format='svg', dpi=1200)
plt.show()

print('Saved: exports/heatmap_1200dpi.svg')
````

**Note on adapting the snippet:** it ran as generated with only the DataFrame swapped in — but worth flagging that `dpi=1200` has no real effect on an SVG. SVG is a vector format (shapes/paths, not a pixel grid), so it renders crisply at any zoom level regardless of the `dpi` argument; that setting only matters for raster formats like PNG (as in Tasks 1 and 3). The AI-suggested `dpi=1200` here is harmless but not doing anything for SVG output specifically.
