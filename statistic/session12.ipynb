# Session 12 - Hypothesis Testing Concepts

## Task 1 — Hypotheses for a Zomato Cashback Offer

**Null hypothesis (H₀):** The cashback offer has no effect on the average number of daily orders — the average after the offer is the same as before (μ_after = μ_before).

**Alternative hypothesis (H₁):** The cashback offer has increased the average number of daily orders (μ_after > μ_before).

This is a **one-tailed** test since the claim specifically is that orders *increased*, not just that they changed in either direction.

## Task 2 — Which Error Type: Wrongly Concluding the Algorithm Works

Wrongly concluding the recommendation algorithm increased listening time when it actually didn't means the null hypothesis (no real change) is true but got rejected anyway — that's a **Type I error**.

**Why:** a Type I error is specifically defined as rejecting a true null hypothesis (a false positive), which is exactly the mistake of declaring the algorithm effective when, in reality, listening time didn't genuinely improve.

## Task 3 — p = 0.03 at 95% Confidence: Reject or Fail to Reject?

At a 95% confidence level, the significance threshold is α = 0.05.

```python
p_value = 0.03
alpha = 0.05

decision = 'Reject the null hypothesis' if p_value < alpha else 'Fail to reject the null hypothesis'
print(f'p-value = {p_value}, alpha = {alpha}')
print(decision)
```

**In context:** since p (0.03) < α (0.05), you **reject the null hypothesis**. That means there's statistically significant evidence that the new discount banner increased purchases — if the banner truly had no effect, you'd expect to see a difference this large (or larger) only 3% of the time by chance. That's below the 5% bar the team agreed on, so it's reasonable for Flipkart to conclude the banner is working and consider rolling it out more broadly — while keeping in mind that 'statistically significant' doesn't automatically mean the increase is large enough to matter commercially.

## Task 4 — What a 99% Confidence Level Means for Push Notifications

Choosing 99% confidence (α = 0.01) instead of the more common 95% means you're demanding much stronger evidence before concluding push notifications genuinely increase engagement. Practically, it means: if this same test were repeated many times, the resulting confidence interval would capture the true effect 99% of the time — and you'd only reject the null (conclude the alerts help) if the p-value comes in below 0.01, versus the more lenient 0.05 threshold.

**Effect on the decision:** this raises the bar for accepting the feature. It sharply reduces the chance of a **Type I error** — falsely rolling out alerts that don't actually help engagement — which matters if a wrong launch is costly (e.g., alert fatigue driving uninstalls). But it comes at a cost: a genuinely helpful feature with a modest effect size is more likely to fail to clear this stricter bar, especially without a larger sample size, increasing the risk of a **Type II error** — rejecting a feature that actually works. Choosing 99% is essentially a deliberate trade-off: fewer false positives, at the cost of being more likely to miss a real but smaller improvement.
