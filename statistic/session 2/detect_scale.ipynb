```python
def detect_scale(column):
    values = [v for v in column if v is not None]
    if not values:
        return "Unknown"

    is_numeric = all(isinstance(v, (int, float)) for v in values)

    if not is_numeric:
        ordinal_words = {"low", "medium", "high", "poor", "average",
                         "good", "excellent", "placed", "shipped",
                         "out for delivery", "delivered", "cancelled"}
        seen = {str(v).strip().lower() for v in values}
        if seen & ordinal_words:
            return "Ordinal"
        return "Nominal"

    has_negative = any(v < 0 for v in values)
    if has_negative:
        return "Interval"

    is_whole_numbers = all(float(v).is_integer() for v in values)
    subtype = "Discrete" if is_whole_numbers else "Continuous"
    return f"Ratio ({subtype})"
```
