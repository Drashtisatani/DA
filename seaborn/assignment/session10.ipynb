```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(8)
```

## Task 1 — 'poster' Context on a Spotify Streams Barplot

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
streams = [1200, 1350, 1280, 1420, 1600, 1900, 1750]
spotify_df = pd.DataFrame({'day': days, 'streams': streams})

sns.set_theme(context='notebook', style='whitegrid')
plt.figure(figsize=(7, 4.5))
sns.barplot(data=spotify_df, x='day', y='streams')
plt.title('Default context: notebook')
plt.tight_layout()
plt.show()

sns.set(context='poster')
plt.figure(figsize=(7, 4.5))
sns.barplot(data=spotify_df, x='day', y='streams')
plt.title('Poster context')
plt.tight_layout()
plt.show()

sns.set_theme(context='notebook', style='whitegrid')  # reset
```

**Observed change:** the `poster` context scales up nearly everything — font sizes for the title, axis labels and tick labels, plus line/bar widths — designed for a plot that'll be viewed from across a room, unlike the default `notebook` context sized for on-screen reading.

## Task 2 — Countplot with 'colorblind' Palette: Food Orders by Cuisine

```python
cuisines = np.random.choice(['Indian', 'Chinese', 'Italian'], 150, p=[0.5, 0.3, 0.2])
cuisine_df = pd.DataFrame({'cuisine': cuisines})

plt.figure(figsize=(7, 4.5))
sns.countplot(data=cuisine_df, x='cuisine', hue='cuisine', palette='colorblind', legend=False)
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.title('Zomato-style Food Orders by Cuisine (colorblind palette)')
plt.tight_layout()
plt.show()
```

## Task 3 — Lineplot: 2 Weeks of Steps, 'muted' Palette + 'notebook' Context

```python
sns.set_theme(context='notebook', style='whitegrid', palette='muted')

steps_2weeks = np.random.normal(7500, 1500, 14).round().astype(int).clip(2000, None)
steps_df = pd.DataFrame({'day': range(1, 15), 'steps': steps_2weeks})

plt.figure(figsize=(8, 4.5))
sns.lineplot(data=steps_df, x='day', y='steps', marker='o')
plt.xlabel('Day')
plt.ylabel('Daily Steps')
plt.title('Daily Step Counts — 2 Weeks (muted palette, notebook context)')
plt.tight_layout()
plt.show()
```

## Task 4 — Refactor: 'deep'→'colorblind' Palette, 'talk'→'poster' Context

```python
category_sales = pd.DataFrame({
    'category': ['Electronics', 'Fashion', 'Home', 'Grocery', 'Beauty'],
    'sales': [4200, 3100, 2800, 3900, 2200],
})

sns.set_theme(context='talk', style='whitegrid', palette='deep')
plt.figure(figsize=(8, 5))
sns.barplot(data=category_sales, x='category', y='sales', hue='category', legend=False)
plt.title("Before: palette='deep', context='talk'")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()

sns.set_theme(context='poster', style='whitegrid', palette='colorblind')
plt.figure(figsize=(8, 5))
sns.barplot(data=category_sales, x='category', y='sales', hue='category', legend=False)
plt.title("After: palette='colorblind', context='poster'")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()

sns.set_theme(context='notebook', style='whitegrid')  # reset
```

**Visual differences noticed:** switching context from `talk` to `poster` makes all text and bar elements noticeably larger and bolder, suited for a big screen or printed poster rather than a slide. Switching the palette from `deep` to `colorblind` shifts the actual bar colors — `colorblind` uses a set specifically chosen to stay distinguishable for people with common forms of color vision deficiency, which `deep` doesn't guarantee.

## Task 5 — Flipkart-Dashboard-Style Theme: 'deep' Palette, 'talk' Context, Custom Font Scale

```python
sns.set_theme(
    context='talk',
    style='whitegrid',
    palette='deep',
    font_scale=1.1,
)

product_categories = pd.DataFrame({
    'category': ['Mobiles', 'Laptops', 'Fashion', 'Home & Kitchen', 'Books'],
    'sales': [58000, 41000, 36000, 27000, 9500],
})

plt.figure(figsize=(9, 5.5))
sns.barplot(data=product_categories, x='category', y='sales', hue='category', legend=False)
plt.xlabel('Product Category')
plt.ylabel('Units Sold')
plt.title('Flipkart-style Sales Dashboard: Top 5 Categories')
plt.tight_layout()
plt.show()

sns.set_theme(context='notebook', style='whitegrid')  # reset for later cells
```
