```python
import sqlite3
import pandas as pd

# Step 1 - connect
conn = sqlite3.connect("orders.db")

# Step 2 - query using GROUP BY + HAVING
query = """
    SELECT user_id, COUNT(order_id) AS total_orders
    FROM orders
    GROUP BY user_id
    HAVING COUNT(order_id) > 5
"""

# Step 3 - load into DataFrame
df = pd.read_sql(query, conn)
conn.close()

# Step 4 - display
print(df)
```
