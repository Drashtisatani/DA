```python
import sqlite3
import pandas as pd

# Connect to SQLite database
conn = sqlite3.connect("movies.db")

# Create table
conn.execute("""
CREATE TABLE IF NOT EXISTS movies (
    id INTEGER PRIMARY KEY,
    title TEXT,
    rating REAL
)
""")

# Clear old data
conn.execute("DELETE FROM movies")

# Insert sample data
conn.execute("INSERT INTO movies (title, rating) VALUES ('Inception', 8.8)")
conn.execute("INSERT INTO movies (title, rating) VALUES ('Avatar', 7.9)")
conn.execute("INSERT INTO movies (title, rating) VALUES ('The Dark Knight', 9.0)")
conn.execute("INSERT INTO movies (title, rating) VALUES ('Titanic', 7.8)")
conn.execute("INSERT INTO movies (title, rating) VALUES ('Interstellar', 8.7)")

conn.commit()

# Required query
query = """
SELECT *
FROM movies
WHERE rating > 8
"""

df = pd.read_sql(query, conn)

print(df)

conn.close()
```
