```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_theme(style='whitegrid')
np.random.seed(1)
```

## Task 1 — Countplot: Swiggy Orders by Payment Method

```python
payment_methods = np.random.choice(['UPI', 'Card', 'Cash'], size=150, p=[0.55, 0.30, 0.15])
orders_df = pd.DataFrame({'payment_method': payment_methods})

plt.figure(figsize=(6, 4.5))
sns.countplot(data=orders_df, x='payment_method', order=['UPI', 'Card', 'Cash'])
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.title('Swiggy Orders by Payment Method')
plt.tight_layout()
plt.show()
```

## Task 2 — Barplot: Average Delivery Time by Food Category

```python
categories = ['Pizza', 'Burger', 'Biryani', 'South Indian']
food_rows = []
base_times = {'Pizza': 32, 'Burger': 22, 'Biryani': 38, 'South Indian': 27}
for cat in categories:
    for _ in range(25):
        food_rows.append({'category': cat, 'delivery_time': base_times[cat] + np.random.normal(0, 5)})
food_df = pd.DataFrame(food_rows)

plt.figure(figsize=(7, 4.5))
sns.barplot(data=food_df, x='category', y='delivery_time', estimator=np.mean)
plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (min)')
plt.title('Average Delivery Time by Food Category')
plt.tight_layout()
plt.show()
```

## Task 3 — Same Barplot with estimator=sum (Total Orders)

```python
plt.figure(figsize=(7, 4.5))
sns.barplot(data=food_df, x='category', y='delivery_time', estimator=np.sum)
plt.xlabel('Food Category')
plt.ylabel('Sum of Delivery Times (min)')
plt.title('Total Delivery Time by Food Category (estimator=sum)')
plt.tight_layout()
plt.show()
```

## Task 4 — Countplot of Movie Genres, No Confidence Interval

```python
genres = np.random.choice(
    ['Action', 'Comedy', 'Drama', 'Thriller', 'Romance'],
    size=120, p=[0.30, 0.20, 0.20, 0.15, 0.15]
)
movies_df = pd.DataFrame({'genre': genres})

plt.figure(figsize=(7, 4.5))
sns.countplot(data=movies_df, x='genre')
plt.xlabel('Genre')
plt.ylabel('Number of Movies')
plt.title('BookMyShow-style Movie Genre Counts')
plt.tight_layout()
plt.show()
```

```python
genre_counts = movies_df['genre'].value_counts().reset_index()
genre_counts.columns = ['genre', 'count']

plt.figure(figsize=(7, 4.5))
sns.barplot(data=genre_counts, x='genre', y='count', errorbar=None)
plt.xlabel('Genre')
plt.ylabel('Number of Movies')
plt.title('Movie Genre Counts — barplot with no confidence interval')
plt.tight_layout()
plt.show()
```
