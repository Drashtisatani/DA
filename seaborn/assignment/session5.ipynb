```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
tips = sns.load_dataset('tips')
print(tips.shape)
print(tips.head())
```

## Task 1 — Basic Pairplot: All Numerical Columns

```python
g = sns.pairplot(tips)
g.figure.suptitle('Pairplot: tips dataset (numerical columns)', y=1.02)
plt.show()
```

## Task 2 — Pairplot with hue='sex'

```python
g = sns.pairplot(tips, hue='sex')
g.figure.suptitle("Pairplot: tips dataset, hue='sex'", y=1.02)
plt.show()
```

With `hue='sex'`, each point is colored by gender, so the scatter panels now show whether male and female customers cluster differently on things like `total_bill` vs `tip` — and the diagonal switches from a single histogram per variable to one overlaid distribution per gender, making it easy to compare their spread side by side rather than just seeing the combined distribution.

## Task 3 — Pairplot with hue='smoker' and diag_kind='kde'

```python
g = sns.pairplot(tips, hue='smoker', diag_kind='kde')
g.figure.suptitle("Pairplot: tips dataset, hue='smoker', diag_kind='kde'", y=1.02)
plt.show()
```

## Task 4 — Pairplot: Selected Columns Only, hue='time'

```python
g = sns.pairplot(tips, vars=['total_bill', 'tip'], hue='time')
g.figure.suptitle("Pairplot: total_bill & tip, hue='time'", y=1.05)
plt.show()
```
