# Session 15 - ANOVA (Analysis of Variance)


```python
import pandas as pd
import numpy as np
from scipy.stats import f_oneway
```

## Task 2 — One-Way ANOVA on purchase_value Across Age Groups

```python
df = pd.read_csv('purchase_data.csv')
print(df)

groups = [g['purchase_value'].values for _, g in df.groupby('age_group')]
f_stat, p_value = f_oneway(*groups)

print('\nF-statistic:', f_stat)
print('p-value:', p_value)
```

## Task 3 — Interpretation

**Answer:** With F ≈ 80.0 and p ≈ 1.15e-07 (far below α = 0.05), there **is** a statistically significant difference in average purchase value across the three age groups. The 26-35 group's average purchase value stands out as clearly higher than the other two, and that gap is too large to be explained by random sampling variation alone.

## Task 4 — Between-Group and Within-Group Variance

Between-group variance is calculated as the mean square between groups (**MS_between** = SS_between / (k-1)); within-group variance is the mean square within groups (**MS_within** = SS_within / (N-k)). Their ratio is exactly the F-statistic from Task 2, which is a useful cross-check.

```python
overall_mean = df['purchase_value'].mean()
k = df['age_group'].nunique()
N = len(df)

group_stats = df.groupby('age_group')['purchase_value']

ss_between = sum(len(g) * (g.mean() - overall_mean) ** 2 for _, g in group_stats)
ss_within = sum(((g - g.mean()) ** 2).sum() for _, g in group_stats)

df_between = k - 1
df_within = N - k

between_group_variance = ss_between / df_between
within_group_variance = ss_within / df_within

print('Between-group variance (MS_between):', between_group_variance)
print('Within-group variance (MS_within):', within_group_variance)
print('F cross-check (between/within):', between_group_variance / within_group_variance)
```

**What each represents:** the **between-group variance** (~1,602,207) measures how much the three age groups' *average* purchase values differ from the overall average — a big number here means the group means are spread far apart. The **within-group variance** (~20,023) measures how much purchase values vary *inside* each age group around their own group's mean — a small number here means customers within the same age group spend fairly similarly. ANOVA compares these two: when between-group variance dwarfs within-group variance (as it does here, by a factor of ~80), it's strong evidence the group means are genuinely different, not just randomly scattered.

## Task 5 — AI-Generated ANOVA Code: Zomato Customer Segments

**Prompt used:** *'Write Python code using scipy.stats to perform a one-way ANOVA comparing average monthly order amounts across three Zomato customer segments: Students, Working Professionals, and Retirees.'*

**Code (as generated, run below on dummy monthly order-amount data for each segment):**

```python
students = [420, 380, 450, 400, 390, 410, 370, 430]
working_professionals = [1200, 1350, 1100, 1280, 1420, 1150, 1300, 1250]
retirees = [700, 650, 720, 680, 690, 710, 660, 705]

f_stat, p_value = f_oneway(students, working_professionals, retirees)

print('Students mean:', np.mean(students))
print('Working Professionals mean:', np.mean(working_professionals))
print('Retirees mean:', np.mean(retirees))
print('\nF-statistic:', f_stat)
print('p-value:', p_value)
```

**Output interpretation:** F ≈ 365.1, p ≈ 4.87e-17 — extremely significant. Average monthly order amounts clearly differ across the three segments (Working Professionals ~₹1256, Retirees ~₹689, Students ~₹406), consistent with each group having distinctly different spending capacity and ordering habits on the platform.
